# Question 1 — E-Commerce Order Priority Management

```python
## Business Scenario

# An e-commerce company maintains a list containing the **order values of orders waiting for processing**.

# During the day:

#   * New orders are received.
#   * Bulk orders are imported from another system.
#   * A VIP order must be placed at a specific position.
#   * A cancelled order must be removed.
#   * The operations team wants to identify an order's position.
#   * Duplicate order amounts must be counted.
#   * Orders have to be sorted by value.
#   * A backup must be created before clearing the end-of-day processing list.

### Sample Data
orders = [1200, 850, 1500, 600, 850]
new_orders = [2000, 750, 1100]

## Functions to Implement

def add_order(orders, order_amount):
    orders.append(order_amount)

def add_bulk_orders(orders, new_orders):
    orders.extend(new_orders)

def add_vip_order(orders, position, order_amount):
    orders.insert(position, order_amount)

def cancel_order(orders, order_amount):
    orders.remove(order_amount)

def remove_last_order(orders):
    return orders.pop()

def find_order_position(orders, order_amount):
    return orders.index(order_amount)

def count_order_amount(orders, order_amount):
    return orders.count(order_amount)

def sort_orders(orders):
    orders.sort()

def reverse_orders(orders):
    orders.reverse()

def backup_orders(orders):
    return orders.copy()

def close_day(orders):
    orders.clear()


add_order(orders, 950)
print(f"{orders=}")

add_bulk_orders(orders, new_orders)
print(f"{orders=}")

add_vip_order(orders, 1, 5000)
print(f"{orders=}")

cancel_order(orders, 600)
print(f"{orders=}")

removed_last_order = remove_last_order(orders)
print(f"{removed_last_order=}")

order_position = find_order_position(orders, 1500)
print(f"{order_position=}")

count_order_amount = count_order_amount(orders, 850)
print(f"{count_order_amount=}")

sort_orders(orders)
print(f"{orders=}")

reverse_orders(orders)
print(f"{orders=}")

backup_orders = backup_orders(orders)
print(f"{backup_orders=}")

close_day(orders)
print(f"{orders=}")
print(f"{backup_orders=}")

# ## Requirements

#   Perform the following business operations:

#   1. Add an order worth `950`.
#   2. Add bulk orders `[2000, 750, 1100]`.
#   3. Insert a VIP order worth `5000` at position `1`.
#   4. Cancel the order having value `600`.
#   5. Remove the last order from the processing queue.
#   6. Find the position of order value `1500`.
#   7. Count how many orders have value `850`.
#   8. Sort orders in ascending order.
#   9. Reverse the sorted order.
#   10. Create a backup copy.
#   11. Clear the original list after processing.
```

# Question 2 — Hospital Patient Queue Management

```python
## Business Scenario

# A hospital reception desk maintains a Python List containing the names of patients waiting for consultation.

#   The reception system must support:

#   * Registering a new patient.
#   * Adding patients transferred from another department.
#   * Giving an emergency patient priority.
#   * Cancelling a registration.
#   * Calling the next patient.
#   * Finding a patient's queue position.
#   * Detecting duplicate registrations.
#   * Alphabetically arranging patients.
#   * Reversing the display order.
#   * Creating a backup before resetting the queue.

### Sample Data

patients = [
    "Arun",
    "Meena",
    "Rahul",
    "Divya",
    "Meena"
]

transferred_patients = [
    "Karthik",
    "Priya",
    "John"
]


## Functions to Implement

def register_patient(patients, patient_name):
    patients.append(patient_name)

def add_transferred_patients(patients, transferred_patients):
    patients.extend(transferred_patients)

def add_emergency_patient(patients, position, patient_name):
    patients.insert(position, patient_name)

def cancel_registration(patients, patient_name):
    patients.remove(patient_name)

def call_last_patient(patients):
    return patients.pop()

def find_patient_position(patients, patient_name):
    return patients.index(patient_name)

def count_registration(patients, patient_name):
    return patients.count(patient_name)

def arrange_patients(patients):
    patients.sort()

def reverse_patient_list(patients):
    patients.reverse()

def create_patient_backup(patients):
    return patients.copy()

def reset_patient_queue(patients):
    patients.clear()

register_patient(patients, "Suresh")
print(f"{patients=}")

add_transferred_patients(patients, transferred_patients)
print(f"{patients=}")

add_emergency_patient(patients, 0, "Emergency-Ravi")
print(f"{patients=}")

cancel_registration(patients, "Rahul")
print(f"{patients=}")

removed_patient = call_last_patient(patients)
print(f"{removed_patient=}")

patient_position = find_patient_position(patients, "Divya")
print(f"{patient_position=}")

patient_count = count_registration(patients, "Meena")
print(f"{patient_count=}")

arrange_patients(patients)
print(f"{patients=}")

reverse_patient_list(patients)
print(f"{patients=}")

patients_backup = create_patient_backup(patients)
print(f"{patients_backup=}")

reset_patient_queue(patients)
print(f"{patients=}")
print(f"{patients_backup=}")

  # ## Requirements

  # Using the given data:

  # 1. Register `"Suresh"`.
  # 2. Add transferred patients `"Karthik"`, `"Priya"` and `"John"`.
  # 3. Insert emergency patient `"Emergency-Ravi"` at position `0`.
  # 4. Cancel `"Rahul"`'s registration.
  # 5. Remove the last patient from the queue.
  # 6. Find the position of `"Divya"`.
  # 7. Count how many times `"Meena"` occurs.
  # 8. Sort the patient names alphabetically.
  # 9. Reverse the patient list.
  # 10. Create a backup.
  # 11. Clear the original queue.
```

# Question 3 — Logistics Delivery Route Management

```python
## Business Scenario

# A logistics company maintains a list containing the **delivery locations assigned to a delivery vehicle**.

# During route planning:

#   * New delivery locations can be added.
#   * Extra locations may be assigned in bulk.
#   * A priority delivery location can be inserted at a specific point.
#   * Cancelled delivery locations have to be removed.
#   * A location can be removed after completion.
#   * The dispatcher should be able to find a location's position.
#   * Duplicate locations must be identified.
#   * Locations may be sorted alphabetically.
#   * The route may be reversed for return travel.
#   * A copy must be saved before completing the route.

### Sample Data

route = [
    "Chennai",
    "Tambaram",
    "Chengalpattu",
    "Mahabalipuram",
    "Tambaram"
]

additional_stops = [
    "Kalpakkam",
    "Pondicherry"
]


## Functions to Implement

def add_delivery_stop(route, location):
    route.append(location)

def add_multiple_stops(route, additional_stops):
    route.extend(additional_stops)

def add_priority_stop(route, position, location):
    route.insert(position, location)

def cancel_delivery_stop(route, location):
    route.remove(location)

def complete_delivery_stop(route, position):
    return route.pop(position)

def find_delivery_stop(route, location):
    return route.index(location)

def count_delivery_stop(route, location):
    return route.count(location)

def sort_delivery_route(route):
    route.sort()

def reverse_delivery_route(route):
    route.reverse()

def backup_delivery_route(route):
    return route.copy()

def complete_route(route):
    route.clear()

add_delivery_stop(route, "Sholinganallur")
print(f"{route=}")

add_multiple_stops(route, additional_stops)
print(f"{route=}")

add_priority_stop(route, 1, "Hospital-Emergency")
print(f"{route=}")

cancel_delivery_stop(route, "Mahabalipuram")
print(f"{route=}")

removed_delivery_stop = complete_delivery_stop(route, 2)
print(f"{removed_delivery_stop=}")

delivery_stop_position = find_delivery_stop(route, "Chengalpattu")
print(f"{delivery_stop_position=}")

count_delivery_stop = count_delivery_stop(route, "Tambaram")
print(f"{count_delivery_stop=}")

sort_delivery_route(route)
print(f"{route=}")

reverse_delivery_route(route)
print(f"{route=}")

backup_route = backup_delivery_route(route)
print(f"{backup_route=}")

complete_route(route)
print(f"{route=}")
print(f"{backup_route=}")

  # ## Requirements:

  # 1. Add `"Sholinganallur"` as a new stop.
  # 2. Add `"Kalpakkam"` and `"Pondicherry"` in one operation.
  # 3. Insert `"Hospital-Emergency"` at position `1`.
  # 4. Cancel the `"Mahabalipuram"` delivery.
  # 5. Remove the delivery location at position `2`.
  # 6. Find the position of `"Chengalpattu"`.
  # 7. Find how many times `"Tambaram"` occurs.
  # 8. Sort all locations alphabetically.
  # 9. Reverse the route.
  # 10. Save a copy of the route.
  # 11. Clear the route after the vehicle completes all deliveries.
```
