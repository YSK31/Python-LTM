About Project

```python
# Mini Project: Shopping Cart System:
# ************************************
# Business Scenario:
# ****************
# You are a Python Developer working for an online retail store. The store wants to build a basic shopping cart system to help manage products, customers, and their orders.

# The system should be able to:
# ************************
# 	Store available products with their prices (Product Catalog)
# 	Allow customers to add or remove items from their cart
# 	Ensure unique customer accounts (no duplicates allowed)
# 	Keep each customer’s details fixed once registered (ID, Name, Email)
# 	At checkout, calculate the total bill for the customer’s cart

# Sample Dataset:
# **************
# 1. Product Catalog (Dictionary)

# products = {
#     1: ("Laptop", 1200.00),
#     2: ("Smartphone", 600.00),
#     3: ("Headphones", 50.00),
#     4: ("Smartwatch", 200.00)
# }

# 2. Customers (Set to ensure uniqueness)

# customers = {
#     (101, "Dinesh Karthik", "dk@example.com"),
#     (102, "Ajay Aravind", "aa@example.com")
# }

# 3. Example Shopping Cart (List of Tuples)

# cart = [
#     (1, 2),   # 2 Laptops
#     (3, 1)    # 1 Headphones
# ]


# Tasks to be Completed:
# *******************
# 1. Display Available Products -- Show all products with IDs, names, and prices.

# 2. Add an Item to Cart -- A customer can select a product by ID and specify the quantity to add.

# 3. Remove an Item from Cart -- A customer can remove a product from the cart by specifying its ID.

# 4. Prevent Duplicate Customers -- Ensure that when registering new customers, IDs or emails are unique.

# 5. Checkout and Calculate Total Bill -- For each product in the cart, multiply quantity × price and calculate the total bill.
```

Pre-seeded Data

```python
products = {
    1: ("Laptop", 1200.00),
    2: ("Smartphone", 600.00),
    3: ("Headphones", 50.00),
    4: ("Smartwatch", 200.00)
}

customers = {
    (101, "Dinesh Karthik", "dk@example.com"),
    (102, "Ajay Aravind", "aa@example.com")
}

# [1, 2],   # 2 Laptops
# [3, 1]    # 1 Headphones
cart = []
```

## Code

```python
def available_products():
    print("------------\nAvailable Products :-")
    for key,val in products.items():
        print(f"ID: {key}, Name: {val[0]}, Price: {val[1]}")
    print("------------")

def display_cart():
    if not cart:
        print("Cart is empty!")
        return
    print("------------\nCart Items :-")
    for item in cart:
        print(f"Product ID: {item[0]}, Product Name: {products[item[0]][0]}, Quantity: {item[1]}")
    print("------------")

def display_customers():
    if not customers:
        print("No customers registered!")
        return
    print("------------\nRegistered Customers :-")
    for customer in customers:
        print(f"ID: {customer[0]}, Name: {customer[1]}, Email: {customer[2]}")
    print("------------")
    
def add_item_to_cart(id, quantity):
    if id not in products:
        print("Product Id doesn't exist!")
        return
    if quantity <= 0:
        print("Quantity cannot be less than 0!")
        return
    for item in cart:
        if item[0] == id:
            item[1] += quantity
            print(f"Added {quantity} units to product ({id})")
            break
    else:
        cart.append([id, quantity])
        print(f"Added new product ({id}) with {quantity} units")

def remove_item_from_cart(id):
    if id not in products:
        print("Product Id doesn't exist!")
        return
    for i in range(len(cart)):
        if cart[i][0] == id:
            cart.pop(i)
            print(f"Removed product ({id}) from cart")
            break
    else:
        print(f"Product ({id}) not found in cart!")

def register_customer(id, name, email):
    if not name.strip():
        print("Name must not be empty!!")
        return
    if not email.strip():
        print("Email must not be empty!!")
        return
    ids = []
    names = []
    emails = []
    for customer in customers:
        ids.append(customer[0])
        names.append(customer[1])
        emails.append(customer[2])
    if id in ids:
        print(f"ID ({id}) already exist!")
        return
    if email in emails:
        print(f"Email ({email}) already exists!")
        return
    for i in range(len(ids)):
        if ids[i] == id and names[i] == name and emails[i] == email:
            print(f"ID: {id}, Name: {name}, Email: {email} | ALREADY EXISTS!")
            break
    customers.add((id, name, email))
    print(f"ID: {id}, Name: {name}, Email: {email} | Registered...")

def remove_register_customer(id):
    global customers
    list_of_customers = list(customers)
    for i in range(len(list_of_customers)):
        if list_of_customers[i][0] == id:
            list_of_customers.pop(i)
            customers = set(list_of_customers)
            print(f"Removed Customer ({id})..")
            break
    else:
        print(f"Customer ({id}) not found!")

def calculate_total_bill():
    if not cart:
        print("Cart is empty!")
        return
    total_amount = 0
    for item in cart:
        total_amount += item[1] * products[item[0]][1]
    print(f"Total Amount = {total_amount}")

def empty_products():
    products.clear()
    print("Cleared Products..")
    
def empty_cart():
    cart.clear()
    print("Cleared Cart..")

def empty_customers():
    customers.clear()
    print("Cleared Customers..")
```
