Import Modules

```python
import pandas as pd
import numpy as np
```

### Create Custom Data

```python
# Create a dictionary with sample data
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Heidi', 'Ivan', 'Judy'],
    'Age': [24, 27, 22, 32, 29, 45, 30, np.nan, 25, 28],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix', 'New York', 'Chicago', 'Los Angeles', 'Houston', 'Phoenix'],
    'Score': [88, 92, 75, np.nan, 85, 90, 78, 83, 91, 70],
    'Status': ['Active', 'Inactive', 'Active', 'Active', 'Inactive', 'Active', 'Active', 'Inactive', 'Active', 'Active']
}

# Create the DataFrame
df = pd.DataFrame(data)

print("Original DataFrame:")
display(df)
```

## Data Cleaning Demonstration


#### 1. Handling Missing Values: `fillna()`
Replacing `NaN` (Not a Number) values with a specified value (e.g., 0, mean, or a specific string). This can be done directly on the DataFrame or on a specific column.

**Syntax:** `df.fillna(value, inplace=True)` or `df['column'].fillna(value, inplace=True)`

```python
# Create a copy to demonstrate inplace modification without affecting the original 'df'
df_demo = df.copy()

# Fill missing 'Age' values with the mean age
df_demo['Age'] = df_demo['Age'].fillna(df_demo['Age'].mean())

# Fill missing 'Score' values with 0
df_demo['Score'] = df_demo['Score'].fillna(0)

print("DataFrame after filling missing 'Age' with mean and 'Score' with 0 (using reassignment):")
display(df_demo)
```

#### Additional Missing Value Handling: `ffill()` and `bfill()`
These methods are used to propagate the last valid observation forward (`ffill`) or next valid observation backward (`bfill`) to fill `NaN` values.

**Syntax:** `df.ffill(inplace=True)` or `df.bfill(inplace=True)`

```python
# Create a copy to demonstrate ffill (forward fill)
df_ffill = df.copy()

# Sort by 'Age' and 'Score' for a more meaningful ffill
df_ffill_sorted = df_ffill.sort_values(by=['Age', 'Score']).reset_index(drop=True)
print("DataFrame sorted for ffill demonstration:")
display(df_ffill_sorted)

# Forward fill missing values in 'Age' and 'Score' columns
df_ffill_sorted['Age'] = df_ffill_sorted['Age'].ffill()
df_ffill_sorted['Score'] = df_ffill_sorted['Score'].ffill()

print("\nDataFrame after forward filling missing values (ffill using reassignment):")
display(df_ffill_sorted)
```

```python
# Create a copy to demonstrate bfill (backward fill)
df_bfill = df.copy()

# Sort by 'Age' and 'Score' for a more meaningful bfill
df_bfill_sorted = df_bfill.sort_values(by=['Age', 'Score'], ascending=False).reset_index(drop=True)
print("DataFrame sorted (desc) for bfill demonstration:")
display(df_bfill_sorted)

# Backward fill missing values in 'Age' and 'Score' columns
df_bfill_sorted['Age'] = df_bfill_sorted['Age'].bfill()
df_bfill_sorted['Score'] = df_bfill_sorted['Score'].bfill()

print("\nDataFrame after backward filling missing values (bfill using reassignment):")
display(df_bfill_sorted)
```

#### Identifying Missing Values: `isna()` and `isnull()`
Both `df.isna()` and `df.isnull()` are aliases and return a boolean DataFrame of the same shape as the original DataFrame, indicating `True` where values are `NaN` (or `None` for object types) and `False` otherwise. They are very useful for identifying and counting missing values.

**Syntax:** `df.isna()` or `df.isnull()`

```python
# Identify missing values using isna()
print("Boolean DataFrame indicating missing values (using isna()):")
display(df.isna())

# Count missing values per column using isna().sum()
print("\nNumber of missing values per column (using isna().sum()):")
display(df.isna().sum())

# Identify missing values using isnull() (alias for isna())
print("\nBoolean DataFrame indicating missing values (using isnull()):")
display(df.isnull())

# Count missing values per row using isnull().sum(axis=1)
print("\nNumber of missing values per row (using isnull().sum(axis=1)):")
display(df.isnull().sum(axis=1))
```

#### 2. Handling Missing Values: `dropna()`
Removing rows or columns that contain missing values.

**Syntax:** `df.dropna(axis=0, how='any', inplace=False)`
*   `axis`: This parameter determines whether to drop rows or columns.
    *   `axis=0` (default): Refers to rows. The operation is applied row-wise.
    *   `axis=1`: Refers to columns. The operation is applied column-wise.
*   `how`: 'any' (drop if any NaN is present) or 'all' (drop if all values are NaN).

```python
# Create a copy to demonstrate different dropna scenarios without altering the original df
df_demo = df.copy()

# Drop rows with any missing values (axis=0, how='any' are defaults)
print("DataFrame after dropping rows with any missing values:")
df_rows_any = df_demo.dropna(axis=0, how='any')
display(df_rows_any)

# To demonstrate how='all', let's create a row that is entirely NaN
df_nan_row = df.copy()
df_nan_row.loc[len(df_nan_row)] = [np.nan, np.nan, np.nan, np.nan, np.nan] # Add a completely NaN row
print("\nDataFrame with an all-NaN row:")
display(df_nan_row)

print("DataFrame after dropping rows where ALL values are missing (how='all'):")
df_rows_all = df_nan_row.dropna(axis=0, how='all')
display(df_rows_all)

# Drop columns with any missing values
print("\nDataFrame after dropping columns with any missing values:")
df_cols_any = df_demo.dropna(axis=1, how='any')
display(df_cols_any)

# Demonstrate inplace=True for dropping rows with any missing values
df_inplace_demo = df.copy()
print("\nOriginal DataFrame for inplace dropna demo:")
display(df_inplace_demo)
df_inplace_demo.dropna(inplace=True)
print("DataFrame after dropping rows with any missing values (inplace=True):")
display(df_inplace_demo)
```

#### 3. Data Type Conversion: `astype()`
Changing the data type of a column.

**Syntax:** `df['column'].astype(dtype)`

```python
# Convert 'Age' column to integer type (after filling NaN if any, or using errors='ignore' for safety)
df_converted = df.copy()
df_converted['Age'] = df_converted['Age'].fillna(df_converted['Age'].mean()).astype(int)

print("DataFrame with 'Age' column converted to integer type:\n")
display(df_converted.info())
```

#### 3.1 Clamping Values: `clip()`

The `clip()` function is used to trim values at input thresholds. Values outside the interval `[lower, upper]` are clipped to the interval edges. That is, values smaller than `lower` become `lower`, and values larger than `upper` become `upper`.

**Syntax:** `df['column'].clip(lower=min_value, upper=max_value)`

```python
# Create a copy to demonstrate clip()
df_clip = df.copy()

# Let's say we want to ensure 'Age' is between 20 and 40, and 'Score' is between 60 and 100.
# First, we need to fill NaN values in 'Score' and 'Age' before clipping, as clip doesn't handle NaN directly.
df_clip['Age'] = df_clip['Age'].fillna(df_clip['Age'].mean())
df_clip['Score'] = df_clip['Score'].fillna(df_clip['Score'].mean())

print("Original DataFrame (after NaN filled) for clip demonstration:")
display(df_clip)

# Clip 'Age' values to be between 20 and 40
df_clip['Age'] = df_clip['Age'].clip(lower=20, upper=40)

# Clip 'Score' values to be between 60 and 100
df_clip['Score'] = df_clip['Score'].clip(lower=60, upper=100)

print("\nDataFrame after clipping 'Age' (20-40) and 'Score' (60-100):")
display(df_clip)
```

#### 4. Handling Duplicates

Identifying Duplicates: `duplicated()`
This function returns a Boolean Series indicating whether each row is a duplicate of a previous row. It's useful for inspecting duplicates before dropping them.

**Syntax:** `df.duplicated(subset=None, keep='first')`
*   `subset`: Column(s) to consider for identifying duplicates.
*   `keep`: 'first' (mark all but the first occurrence as `True`), 'last' (mark all but the last as `True`), or `False` (mark all duplicates as `True`).

```python
# Create a DataFrame with some duplicate rows for demonstration
df_dup = pd.DataFrame({
    'ColA': [1, 2, 1, 3, 2],
    'ColB': ['X', 'Y', 'X', 'Z', 'Y']
})
print("Original DataFrame with duplicates:")
display(df_dup)

# Using the DataFrame with duplicates from the previous example
print("Original DataFrame with duplicates:")
display(df_dup)

# Identify duplicate rows (default: marks all but the first occurrence as True)
dup_rows_default = df_dup.duplicated()
print("\nBoolean Series identifying duplicate rows (default - marks all but first as True):")
display(dup_rows_default)

# Identify all occurrences of duplicates (marks all duplicate rows as True)
dup_rows_all = df_dup.duplicated(keep=False)
print("\nBoolean Series identifying ALL duplicate rows (keep=False):")
display(dup_rows_all)

# Identify duplicates based on 'ColA' only
dup_col_a = df_dup.duplicated(subset=['ColA'])
print("\nBoolean Series identifying duplicate rows based on 'ColA':")
display(dup_col_a)
```

Removing Duplicates: `drop_duplicates()`
Eliminating duplicate rows from the DataFrame.

**Syntax:** `df.drop_duplicates(subset=None, keep='first', inplace=False)`
*   `subset`: Column(s) to consider for identifying duplicates.
*   `keep`:
    *   `'first'` (default): Keep the first occurrence, drop duplicates.
    *   `'last'`: Keep the last occurrence, drop duplicates.
    *   `False`: Drop all occurrences where a row is duplicated.
*   `inplace`: If `True`, modifies the DataFrame directly.

```python
# Create a DataFrame with some duplicate rows for demonstration
df_extended_dup = pd.DataFrame({
    'ColA': [1, 2, 1, 3, 2, 1],
    'ColB': ['X', 'Y', 'X', 'Z', 'Y', 'X']
})
print("Original DataFrame with duplicates:")
display(df_extended_dup)

# Drop duplicate rows (keeps the first occurrence by default, or explicitly with keep='first')
df_no_dup_first = df_extended_dup.drop_duplicates(keep='first')
print("\nDataFrame after dropping duplicates (keep='first' - default behavior):")
display(df_no_dup_first)

# Drop duplicate rows based on 'ColA' and keep the last occurrence
df_no_dup_subset_last = df_extended_dup.drop_duplicates(subset=['ColA'], keep='last')
print("\nDataFrame after dropping duplicates based on 'ColA', keeping the last:")
display(df_no_dup_subset_last)

# Drop all occurrences of duplicate rows (keep=False)
df_no_dup_all = df_extended_dup.drop_duplicates(keep=False)
print("\nDataFrame after dropping ALL occurrences of duplicate rows (keep=False):")
display(df_no_dup_all)

# Demonstrate inplace=True for dropping duplicates (keeping first as default)
df_inplace_dup = df_extended_dup.copy()
print("\nOriginal DataFrame for inplace drop_duplicates demo:")
display(df_inplace_dup)
def_1 = df_inplace_dup.drop_duplicates(inplace=False)
print("DataFrame after dropping duplicates (inplace=True, keep='first' implicit):")
display(df_inplace_dup)
print(def_1)
```

#### 5. Renaming Columns: `rename()`
Changing the names of columns or index labels.

**Syntax:** `df.rename(columns={'old_name': 'new_name'}, inplace=False)`

```python
# Rename 'Score' to 'Exam_Score' directly on a copy of the DataFrame
df_renamed = df.copy()
df_renamed.rename(columns={'Score': 'Exam_Score'}, inplace=True)

print("DataFrame after renaming 'Score' to 'Exam_Score' (using inplace=True):")
display(df_renamed)
```

#### 6. Applying a Function to a Column: `apply()`
Applying a custom function to a column's values.

**Syntax:** `df['column'].apply(function)`

```python
# Create a new column 'Age_Group' based on 'Age'
df_apply = df.copy()

def get_age_group(age):
    if pd.isna(age):
        return 'Unknown'
    elif age < 30:
        return 'Young'
    else:
        return 'Adult'

df_apply['Age_Group'] = df_apply['Age'].apply(get_age_group)

print("DataFrame with new 'Age_Group' column:")
display(df_apply)
```
