````python
# # Business Scenario: Retail Sales Performance Analysis Using Pandas Dataframe

# ## Scenario

# You are working as a **Junior Data Analyst for an online retail company**. The company has collected sales transaction data from different cities and product categories.

# The management wants you to use **Pandas DataFrame** to:

# - Create and inspect the dataset.
# - Understand DataFrame attributes.
# - Access rows and columns.
# - Apply `loc` and `iloc`.
# - Filter business data.
# - Calculate aggregate values.
# - Create calculated columns.
# - Remove unnecessary columns.
# - Perform DataFrame inspection.
# - Analyze category frequency.
# - Sort sales records.

# This exercise is designed so trainees perform the operations **one after another on the same DataFrame**.

# ---

# # Sample Data

# Use the following data:

import pandas as pd

data = {
    "OrderID": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008],
    "Product": [
        "Laptop",
        "Mouse",
        "Keyboard",
        "Monitor",
        "Laptop",
        "Mouse",
        "Monitor",
        "Keyboard"
    ],
    "Category": [
        "Electronics",
        "Accessories",
        "Accessories",
        "Electronics",
        "Electronics",
        "Accessories",
        "Electronics",
        "Accessories"
    ],
    "City": [
        "Chennai",
        "Bangalore",
        "Chennai",
        "Hyderabad",
        "Bangalore",
        "Chennai",
        "Bangalore",
        "Hyderabad"
    ],
    "Quantity": [2, 5, 3, 2, 1, 4, 3, 2],
    "UnitPrice": [55000, 700, 1200, 15000, 55000, 700, 15000, 1200],
    "Discount": [5000, 100, 200, 1000, 3000, 100, 1500, 150]
}
````

Tasks to be Performed

````python
## Part 1 — DataFrame Creation

# 1. Create a Pandas DataFrame called `sales_df` using the given dictionary.
# 2. Display the complete DataFrame.

sales_df = pd.DataFrame(data)
print(sales_df)
````

````python
# ## Part 2 — DataFrame Attributes

# Display the following:

# 1. Number of rows and columns using `shape`.
# 2. Column names using `columns`.
# 3. Row index using `index`.
# 4. Data type of each column using `dtypes`.
# 5. Underlying DataFrame values using `values`.

print(sales_df.shape)
print(list(sales_df.columns))
print(list(sales_df.index))
print(sales_df.dtypes)
print(list(sales_df.values))
````

````python
# ---

# ## Part 3 — Accessing Data

# 1. Display only the `Product` column.
# 2. Display `Product`, `City`, and `UnitPrice`.
# 3. Display the first row.
# 4. Display the last row.
print(sales_df['Product'])
print(sales_df[['Product', 'City', 'UnitPrice']])
print(sales_df.head(1))
print(sales_df.tail(1))
````

````python

# ---

# ## Part 4 — `loc` Indexing and Slicing

# 1. Display row with index `2` using `loc`.
# 2. Display rows from index `1` to `4` using `loc`.
# 3. Display `Product` and `Quantity` for rows `1` to `4`.
# 4. Display records belonging to Chennai using `loc`.

print(sales_df.loc[2])
print(sales_df.loc[1:4])
print(sales_df[['Product', 'Quantity']].loc[1:4])
print(sales_df.loc[sales_df['City'] == 'Chennai'])
````

````python
# ---

# ## Part 5 — `iloc` Indexing and Slicing

# 1. Display the first row using `iloc`.
# 2. Display the first three rows.
# 3. Display the first three rows and first four columns.
# 4. Display rows from positions `2` to `5`.

print(sales_df.iloc[0])
print(sales_df.iloc[:3])
print(sales_df.iloc[:3,:4])
print(sales_df.iloc[2:5])

````

````python
# ---

# ## Part 6 — Filtering

# 1. Display orders where `Quantity > 2`.
# 2. Display orders where `UnitPrice > 10000`.
# 3. Display only Electronics products.
# 4. Display Chennai orders where quantity is greater than `2`.

print(sales_df[sales_df['Quantity'] > 2])
print(sales_df[sales_df['UnitPrice'] > 10000])
print(sales_df[sales_df['Category'] == "Electronics"])
print(sales_df[(sales_df['Quantity'] > 2) & (sales_df['City'] == "Chennai")])

````

````python
# ---

# ## Part 7 — Aggregate Functions

# 1. Calculate total quantity sold using `sum()`.
# 2. Calculate average quantity sold using `mean()`.
# 3. Calculate total Unit Price using `sum()`.
# 4. Calculate average Unit Price using `mean()`.

print(sales_df['Quantity'].sum())
print(sales_df['Quantity'].mean())
print(sales_df['UnitPrice'].sum())
print(sales_df['UnitPrice'].mean())

````

````python
# ---

# ## Part 8 — Add New Column

# 1. Create a new column called `SalesAmount`:

# ```
# SalesAmount = Quantity × UnitPrice
# ```

# 1. Create another column called `NetAmount`:

# ```
# NetAmount = SalesAmount - Discount
# ```

sales_df['SalesAmount'] = sales_df['Quantity'] * sales_df['UnitPrice']
sales_df['NetAmount'] = sales_df['SalesAmount'] - sales_df['Discount']
print(sales_df)
````

````python
# ---

# ## Part 9 — Drop a Column

# 1. Remove the `Discount` column from the DataFrame.
del sales_df['Discount']
print(sales_df)
````

````python
# ---

# ## Part 10 — DataFrame Inspection

# Perform the following:

# 1. `head()`
# 2. `tail()`
# 3. `info()`
# 4. `describe()`

print(sales_df.head())
print(sales_df.tail())
print(sales_df.info())
print(sales_df.describe())

````

````python
# ---

# ## Part 11 — `value_counts()`

# 1. Find how many orders belong to each category.
# 2. Find the number of orders from each city.
# 3. Find how many times each product was ordered.

print(sales_df['Category'].value_counts())
print(sales_df['City'].value_counts())
print(sales_df['Product'].value_counts())
````

````python
# ---

# ## Part 12 — `sort_values()`

# 1. Sort records according to `Quantity`.
# 2. Sort according to `NetAmount` from highest to lowest.
# 3. Sort using `City` and `NetAmount`.

sorted_quantity_df = sales_df.sort_values(by = 'Quantity')
print(sorted_quantity_df)

sorted_netamount_df = sales_df.sort_values(by = 'NetAmount', ascending = False)
print(sorted_netamount_df)

sorted_city_netamount_df = sales_df.sort_values(by = ['City', 'NetAmount'])
print(sorted_city_netamount_df)
````
