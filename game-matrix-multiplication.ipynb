```python
def create_random_matrix(low, high, size, round_to):
    mat = np.round(np.random.uniform(low, high, size), round_to)
    return mat

def matrix_multiplication(mat1, mat2):
    return mat1.dot(mat2)

def start():
    turn = 0
    while True:
        if turn > 0:
            to_continue = input("Wanna continue (y or n):")
            if to_continue == "n":
                print("bye bye !")
                break
        turn += 1
        print("Start....\n")
        print("-----------\nMatrix :-")
        row_mat_1 = int(input("Matrix-1 Row count:"))
        col_mat_1 = int(input("Matrx-1 Column count:"))
    
        row_mat_2 = int(input("Matrix-2 Row count:"))
        col_mat_2 = int(input("Matrix-2 Column count:"))
    
        if col_mat_1 != row_mat_2:
            print("Rule Violated: Column count of matrix-1 should be equal to Row count of matrix-2 !!!")
            print("So again start...\n")
            continue
            
        print("Random numbers :-")
        low = float(input("Min random value:"))
        high = float(input("Max random value:"))
        round_to = int(input("Round for each random value:"))
    
        mat1 = create_random_matrix(low, high, (row_mat_1, col_mat_1), round_to)
        mat2 = create_random_matrix(low, high, (row_mat_2, col_mat_2), round_to)
        resultant_matrix = matrix_multiplication(mat1, mat2)
        print("\nMatrix-1 :-")
        print(mat1)
        print("\nMatrix-2 :-")
        print(mat2)
        print("\nResultant matrix :-")
        print(resultant_matrix)

    
```

```python
start()
```
